```python
import ollama
```

```python
# read host from llm.env file
with open('llm.env', 'r') as f:
    host = f.read().strip()

host = host.split('=')[1]
print(f"Using Ollama host: {host}")

client = ollama.Client(host=host)
```

```python
response = client.chat(
    model="qwen2.5:7b", 
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
    ]
)
```
