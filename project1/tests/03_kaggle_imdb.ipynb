```python
import kagglehub
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
from tqdm import tqdm
```

```python
df = pd.read_csv('imdb_ratings.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
top_voted = df.sort_values(by='numVotes',ascending=False).head()
import seaborn as sns
plt.figure(figsize=(12,6))
sns.barplot(
    data=top_voted,
    x="numVotes",
    y="title",
    palette="viridis"
)
plt.title("Top 10 Most Voted Movies")
plt.xlabel("Number of Votes")
plt.ylabel("Movie Title")
plt.show()
```

```python
highest_rating  = df[df['numVotes']>1000].sort_values(by='averageRating',ascending=False).head(10)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=highest_rating,
    x="averageRating",
    y="title",
    palette="viridis"
)
plt.title("Top 10 Most Rated Movies")
plt.xlabel("Rating")
plt.ylabel("Movie Title")
plt.xlim(9,10)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df['averageRating'],bins=20,kde=True)
plt.title("Distribution of rating")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df['numVotes'],bins=20,kde=True,log_scale=True)
plt.title("Distribution of votes")
plt.xlabel("Votes")
plt.ylabel("Count")

plt.show()
```
